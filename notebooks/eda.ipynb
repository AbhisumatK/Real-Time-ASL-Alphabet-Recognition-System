# ASL Alphabet Exploratory Data Analysis

This notebook explores the ASL Alphabet dataset, analyzes class distribution, and visualizes samples.

```python
import os
import cv2
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
from src.utils import download_dataset, get_asl_labels

# Download or get path
data_path = download_dataset()
train_path = os.path.join(data_path, 'asl_alphabet_train/asl_alphabet_train')
labels = get_asl_labels()
```

## Class Distribution

```python
stats = []
for label in labels:
    label_dir = os.path.join(train_path, label)
    if os.path.exists(label_dir):
        count = len(os.listdir(label_dir))
        stats.append({'label': label, 'count': count})

df = pd.DataFrame(stats)
plt.figure(figsize=(15, 6))
sns.barplot(data=df, x='label', y='count')
plt.title('Samples per Class')
plt.xticks(rotation=45)
plt.show()
```

## Sample Visualization

```python
fig, axes = plt.subplots(4, 8, figsize=(20, 10))
axes = axes.flatten()

for i, label in enumerate(labels):
    label_dir = os.path.join(train_path, label)
    img_name = os.listdir(label_dir)[0]
    img = cv2.imread(os.path.join(label_dir, img_name))
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    
    axes[i].imshow(img)
    axes[i].set_title(label)
    axes[i].axis('off')

plt.tight_layout()
plt.show()
```
